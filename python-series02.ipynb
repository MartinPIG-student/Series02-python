```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Generate an array of 1000 points between 0 and π
t = np.linspace(0, 1, 1000)

# 2. Calculate the sine of each point
s = np.sin(2 * np.pi * t)

# 3. Create the plot
plt.figure(figsize=(8, 4))
plt.plot(t, s, label="sin(x)", color="blue", linewidth=2)

# 4. print the title and labels
plt.title("Graph of the Sine Function")
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.axis('tight')
plt.grid()
plt.legend()

# 5. Save the figure as a PNG file
plt.savefig("graphique_sinusoide.png", dpi=300, bbox_inches='tight')

# 6. Show the plot
plt.show()


```

```python
# final version 

# A Define sinusoid function 

def sinusoid(t, A, f, phi):
    """
    Calculate the value of a sinusoidal function at given time points.
    
    Parameters:
    t (array): Time points
    A (float): Amplitude
    f (float): Frequency
    phi (float): Phase shift
    
    Returns:
    array: Values of the sinusoidal function at the given time points
    """
    x = A * np.sin(2 * np.pi * f * t + phi)
    return x

# Create s1 and s2 using the sinusoid function
s1 = sinusoid(t, 1, 1, 0)
s2 = sinusoid(t, 1, 2, 0)



```

```python
# plot the cell
plt.figure(figsize=(8, 6))
plt.plot(t, s1, label="Sinus 1")
plt.plot(t, s2, label="Sinus 2")
plt.title("Sinusoidal function")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid()
plt.legend()
plt.show()
```
